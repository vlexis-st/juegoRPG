```python
import pygame
import sys
import os

# ==================== CLASE PERSONAJE ====================
# Qué: Define la estructura de un personaje del juego
# Para qué: Almacenar datos del personaje (clase, agilidad, vitalidad) para uso posterior en el juego
# Cómo: Crea un objeto con 3 atributos principales
# Editable: SÍ - Puedes añadir más atributos (vida, mana, experiencia, etc.) sin afectar la estructura
class Personaje:
    def __init__(self, clase, agilidad, vitalidad, imagen=None):
        # Almacena el nombre de la clase (Clérigo, Guerrero, Mago, Luchador)
        self.clase = clase
        # Almacena el valor de agilidad (afecta velocidad en combate)
        self.agilidad = agilidad
        # Almacena el valor de vitalidad (afecta puntos de vida)
        self.vitalidad = vitalidad
        # Almacena la imagen del personaje (opcional)
        self.imagen = imagen

    def __repr__(self):
        # Método para representar el objeto como string (útil para debug)
        return f"Personaje(clase={self.clase}, agilidad={self.agilidad}, vitalidad={self.vitalidad})"


def cargar_imagen(ruta, tamaño=(120, 120)):
    """
    Carga una imagen desde un archivo y la escala al tamaño especificado.
    
    Parámetros:
        ruta (str): Ruta del archivo de imagen (ej. "imágenes/clerigo.png")
        tamaño (tuple): Tamaño destino (ancho, alto) en píxeles
    
    Retorna:
        pygame.Surface: Imagen escalada o None si no se encuentra el archivo
    
    Ejemplo de uso:
        img = cargar_imagen("imágenes/clerigo.png", (120, 120))
        if img:
            pantalla.blit(img, (x, y))
    """
    try:
        # Intenta cargar la imagen
        imagen = pygame.image.load(ruta)
        # Escala la imagen al tamaño especificado
        imagen = pygame.transform.scale(imagen, tamaño)
        return imagen
    except Exception as e:
        # Si hay error (archivo no existe), imprime el error y retorna None
        print(f"Error cargando imagen '{ruta}': {e}")
        return None


def main():
    # ==================== INICIALIZACIÓN DE PYGAME ====================
    # Qué: Inicia el sistema de gráficos pygame
    # Para qué: Preparar pygame para dibujar en pantalla y procesar eventos
    # Cómo: Llama a pygame.init() que inicializa todos los módulos
    # Editable: NO - Es obligatorio y no debe modificarse
    pygame.init()

    # ==================== CONFIGURACIÓN DE LA VENTANA ====================
    # Qué: Define las dimensiones y propiedades de la ventana del juego
    # Para qué: Establecer el tamaño de la pantalla donde se dibuja todo
    # Cómo: ANCHO=900 y ALTO=600 píxeles (tamaño estándar)
    # Editable: SÍ - Puedes cambiar 900 y 600 a otros valores (ej. 1024, 768) sin afectar la lógica
    ANCHO, ALTO = 900, 600
    
    # Crea la ventana de pygame con las dimensiones especificadas
    # Qué: Crea la superficie principal donde se dibuja
    # Para qué: Obtener un objeto pantalla para dibujar elementos
    # Editable: NO - Necesario para que funcione pygame
    pantalla = pygame.display.set_mode((ANCHO, ALTO))
    
    # Qué: Define el título de la ventana (lo que aparece en la barra superior)
    # Para qué: Identificar la aplicación
    # Cómo: El texto "Juego RPG - Grupo 5" aparece en la barra de título
    # Editable: SÍ - Puedes cambiar el texto sin afectar nada
    pygame.display.set_caption("Juego RPG - Grupo 5")

    # ==================== DEFINICIÓN DE COLORES ====================
    # Qué: Define constantes RGB (Red, Green, Blue) para usar en toda la aplicación
    # Para qué: Evitar escribir (255, 255, 255) cada vez; hace el código legible
    # Cómo: Tuplas (R, G, B) donde cada valor va de 0 a 255
    # Editable: SÍ - Puedes cambiar los valores para obtener colores diferentes
    BLANCO = (255, 255, 255)      # RGB: 255, 255, 255
    NEGRO = (0, 0, 0)              # RGB: 0, 0, 0
    AZUL = (0, 100, 255)           # RGB: Azul medio
    AZUL_CLARO = (100, 150, 255)   # RGB: Azul más claro para hover
    GRIS = (50, 50, 50)            # RGB: Gris oscuro (no se usa actualmente)

    # ==================== DEFINICIÓN DE FUENTES ====================
    # Qué: Define 3 fuentes de texto con diferentes tamaños
    # Para qué: Mostrar títulos, botones y texto general en pantalla
    # Cómo: pygame.font.Font(None, tamaño) - None usa la fuente por defecto del sistema
    # Editable: SÍ - Puedes cambiar los tamaños (72, 48, 36) a otros valores
    # Nota: Tamaños más grandes = texto más grande en pantalla
    fuente_titulo = pygame.font.Font(None, 72)   # Fuente grande para títulos
    fuente_boton = pygame.font.Font(None, 48)    # Fuente mediana para botones
    fuente_texto = pygame.font.Font(None, 36)    # Fuente pequeña para estadísticas

    # ==================== CONFIGURACIÓN DE RELOJ Y FPS ====================
    # Qué: Controla la velocidad del juego (fotogramas por segundo)
    # Para qué: Que el juego corra a velocidad constante en cualquier computadora
    # Cómo: FPS=60 significa 60 fotogramas por segundo (60 veces por segundo actualiza)
    # Editable: SÍ - Puedes cambiar 60 a otro valor (30, 120) pero afecta velocidad
    reloj = pygame.time.Clock()
    FPS = 60

    # ==================== CARGAR IMÁGENES ====================
    # Qué: Carga las imágenes de cada clase desde archivos
    # Para qué: Mostrar imágenes de personajes en los slots
    # Cómo: Busca archivos PNG en la carpeta "imágenes" con nombres por clase
    # Editable: SÍ - Puedes cambiar los nombres de archivo o la carpeta
    # Nota: Asegúrate de que las imágenes existan en la ruta especificada
    
    # Diccionario que almacena las imágenes escaladas de cada clase
    # Qué: {clase: imagen_escalada}
    # Para qué: Acceder rápidamente a la imagen correcta según la clase
    # Editable: SÍ - Puedes cambiar los nombres de las clases o archivos
    imagenes_clases = {
        'clerigo': cargar_imagen("imágenes/clerigo.png", (120, 120)),
        'guerrero': cargar_imagen("imágenes/guerrero.png", (120, 120)),
        'mago': cargar_imagen("imágenes/mago.png", (120, 120)),
        'luchador': cargar_imagen("imágenes/luchador.png", (120, 120)),
    }
    
    # Imagen para slots vacíos (opcional)
    # Qué: Imagen que se muestra en cuadrados sin personaje (en lugar de "+")
    # Para qué: Mostrar una imagen más bonita que el símbolo "+"
    # Editable: SÍ - Puedes cambiar la ruta o eliminar esta línea para usar "+"
    imagen_vacio = cargar_imagen("imágenes/vacio.png", (120, 120))

    # ==================== CLASE BOTÓN ====================
    # Qué: Define un botón reutilizable que se puede usar en cualquier pantalla
    # Para qué: Crear botones interactivos con hover y click
    # Cómo: Almacena posición, tamaño, texto y colores; detecta click e hover
    # Editable: SÍ - Puedes añadir efectos de sonido, animaciones, etc.
    class Boton:
        # Constructor: inicializa un botón en una posición con un texto
        # Parámetros: x, y (posición), ancho, alto (tamaño), texto (label del botón)
        def __init__(self, x, y, ancho, alto, texto):
            # pygame.Rect: estructura que almacena un rectángulo (x, y, ancho, alto)
            # Se usa para detectar colisiones (si el ratón está sobre el botón)
            self.rect = pygame.Rect(x, y, ancho, alto)
            # Almacena el texto que aparecerá en el botón
            self.texto = texto
            # Color base del botón (azul)
            self.color_base = AZUL
            # Color cuando el ratón está sobre el botón (azul más claro)
            self.color_hover = AZUL_CLARO
            # Color actual (comienza con color base)
            self.color_actual = self.color_base

        # Método dibujar: renderiza el botón en la pantalla
        # Qué: Dibuja el botón con su texto en la pantalla
        # Para qué: Mostrar el botón visualmente al usuario
        # Editable: SÍ - Puedes cambiar colores, grosor del borde (3), fuente, etc.
        def dibujar(self, superficie):
            # Dibuja un rectángulo relleno con el color actual
            # El color cambia si hay hover (color_hover) o no (color_base)
            pygame.draw.rect(superficie, self.color_actual, self.rect)
            # Dibuja un borde blanco alrededor del rectángulo (grosor: 3 píxeles)
            # Editable: Puedes cambiar 3 a otro valor para borde más grueso/fino
            pygame.draw.rect(superficie, BLANCO, self.rect, 3)
            # Renderiza el texto del botón
            # Parámetros: (texto, suavizado, color)
            texto_surface = fuente_boton.render(self.texto, True, BLANCO)
            # Calcula la posición para centrar el texto en el botón
            texto_rect = texto_surface.get_rect(center=self.rect.center)
            # Dibuja el texto en la pantalla
            superficie.blit(texto_surface, texto_rect)

        # Método actualizar: detecta si el ratón está sobre el botón (hover)
        # Qué: Cambia el color del botón si el ratón está encima
        # Para qué: Feedback visual para el usuario (sabe que el botón es clickeable)
        # Editable: SÍ - Puedes cambiar los colores de hover
        def actualizar(self, pos_mouse):
            # collidepoint: devuelve True si la posición del ratón está dentro del rectángulo
            if self.rect.collidepoint(pos_mouse):
                # Si está sobre el botón, usa color hover
                self.color_actual = self.color_hover
            else:
                # Si no está sobre el botón, usa color base
                self.color_actual = self.color_base

        # Método fue_clickeado: detecta si el usuario hizo click en el botón
        # Qué: Verifica si el ratón está sobre el botón en el momento del click
        # Para qué: Saber qué botón fue presionado
        # Editable: NO - La lógica es correcta y no necesita cambios
        def fue_clickeado(self, pos_mouse):
            return self.rect.collidepoint(pos_mouse)

    # ==================== MENÚ PRINCIPAL ====================
    # Qué: Crea los dos botones principales del menú
    # Para qué: Permitir al usuario seleccionar "Personaje" o "Salir del juego"
    # Cómo: Crea dos objetos Boton con posiciones y textos específicos
    # Editable: SÍ - Puedes cambiar posiciones, tamaños y textos sin afectar la estructura
    
    # Botón "Seleccionar personaje" en la parte superior
    # Posición: centrado horizontalmente, mitad superior de la pantalla
    # Tamaño: 300 píxeles de ancho, 80 de alto
    # Parámetros editables: posición (ANCHO // 2 - 150), tamaño (300, 80), texto
    boton_personaje = Boton(ANCHO // 2 - 150, ALTO // 2 - 50, 300, 80, "Seleccionar personaje")
    
    # Botón "Salir del juego" en la parte inferior
    # Posición: centrado horizontalmente, mitad inferior de la pantalla
    # Tamaño: 300 píxeles de ancho, 80 de alto
    # Parámetros editables: posición (ALTO // 2 + 100), tamaño (300, 80), texto
    boton_salir = Boton(ANCHO // 2 - 150, ALTO // 2 + 100, 300, 80, "Salir del juego")

    # ==================== PANTALLA A: 6 CUADRADOS CON "+" ====================
    # Qué: Define un layout de 6 cuadrados (3 columnas x 2 filas) para mostrar personajes
    # Para qué: Mostrar slots donde el usuario puede crear hasta 6 personajes
    # Cómo: Calcula posiciones automáticamente para 6 cuadrados distribuidos uniformemente
    # Editable: SÍ - Puedes cambiar tamaño (120, 120), espaciado (30, 30), layout (cols=3, rows=2)
    
    # Tamaño de cada cuadrado (120x120 píxeles)
    # Editable: SÍ - Cambiar estos valores hace cuadrados más grandes/pequeños
    cuadro_w, cuadro_h = 120, 120
    
    # Espaciado entre cuadrados
    # gap_x: espacio horizontal entre cuadrados (30 píxeles)
    # gap_y: espacio vertical entre cuadrados (30 píxeles)
    # Editable: SÍ - Cambiar para más/menos espacio entre cuadrados
    gap_x, gap_y = 30, 30
    
    # Layout: 3 columnas y 2 filas = 6 cuadrados totales
    # Editable: SÍ - Cambiar a cols=2, rows=3 para otro layout (pero cambia la distribución visual)
    cols = 3
    rows = 2
    total_cuadros = cols * rows

    # Lista que almacenará los pygame.Rect de cada cuadrado
    # Qué: Estructura que guarda la posición y tamaño de cada cuadrado
    # Para qué: Detectar clicks en los cuadrados y dibujarlos
    # Editable: NO - Se llena automáticamente en el loop
    cuadros_rects = []
    
    # Calcula el ancho total que ocupan los cuadrados con el espaciado
    # Fórmula: (3 cuadros × 120) + (2 espacios × 30) = 420 píxeles
    # Editable: NO - Se calcula automáticamente basado en cols, cuadro_w, gap_x
    total_width = cols * cuadro_w + (cols - 1) * gap_x
    
    # Calcula la posición X inicial para centrar los cuadrados horizontalmente
    # Qué: Encuentra el punto X donde deben comenzar los cuadrados (centrado)
    # Para qué: Que los 6 cuadrados queden centrados en la pantalla
    # Editable: NO - Se calcula automáticamente
    start_x = ANCHO // 2 - total_width // 2
    
    # Posición Y inicial donde comienzan los cuadrados (140 píxeles desde arriba)
    # Editable: SÍ - Cambiar 140 para mover los cuadrados arriba/abajo
    start_y = 140

    # Loop que crea los 6 cuadrados y calcula sus posiciones
    # Qué: Genera un Rect para cada uno de los 6 cuadrados
    # Para qué: Tener los rectángulos necesarios para detectar clicks y dibujar
    # Editable: NO - La lógica de cálculo es correcta
    for r in range(rows):
        for c in range(cols):
            # Calcula posición X del cuadrado actual (columna c)
            x = start_x + c * (cuadro_w + gap_x)
            # Calcula posición Y del cuadrado actual (fila r)
            y = start_y + r * (cuadro_h + gap_y)
            # Crea un Rect con posición (x, y) y tamaño (cuadro_w, cuadro_h)
            cuadros_rects.append(pygame.Rect(x, y, cuadro_w, cuadro_h))

    # Variable para rastrear sobre qué cuadrado está el ratón (para hover)
    # Qué: Índice del cuadrado donde está el ratón (-1 si ninguno)
    # Para qué: Cambiar color del cuadrado cuando el ratón está encima
    # Editable: NO - Se actualiza automáticamente
    cuadro_hover_index = -1

    # Botón "Volver atrás" en la Pantalla A
    # Qué: Botón para regresar al menú principal desde la pantalla de crear personaje
    # Para qué: Permitir al usuario volver atrás sin crear un personaje
    # Editable: SÍ - Puedes cambiar posición y tamaño
    boton_volver_pantalla_a = Boton(ANCHO // 2 - 150, ALTO - 120, 300, 60, "Volver atrás")

    # ==================== SELECCIÓN DE CLASE ====================
    # Qué: Define el layout para los 4 botones de selección de clase
    # Para qué: Mostrar opciones de clase (Clérigo, Guerrero, Mago, Luchador)
    # Editable: SÍ - Puedes cambiar posiciones, tamaños, orden de clases
    
    # Tamaño de los botones de clase
    # btn_w: 300 píxeles de ancho
    # btn_h: 60 píxeles de alto
    # Editable: SÍ - Cambiar para botones más grandes/pequeños
    btn_w, btn_h = 300, 60
    
    # Posición Y inicial donde empiezan los botones de clase
    # Editable: SÍ - Cambiar 150 para mover botones arriba/abajo
    start_y_clases = ALTO // 2 - 150
    
    # Lista de clases disponibles: (nombre visual, clave interna)
    # Qué: Tuplas con el nombre que se muestra y la clave para buscar stats
    # Para qué: Permitir seleccionar entre 4 clases diferentes
    # Editable: SÍ - Puedes cambiar nombres, pero debes actualizar 'stats' también
    clases = [("Clérigo", 'clerigo'), ("Guerrero", 'guerrero'), ("Mago", 'mago'), ("Luchador", 'luchador')]
    
    # Lista que almacenará los botones de clase
    botones_clases = []
    
    # Loop que crea un botón para cada clase
    # Qué: Genera 4 botones, uno para cada clase
    # Para qué: Tener botones interactivos para seleccionar clase
    # Editable: SÍ - Puedes cambiar espaciado (10) o alineación
    for i, (label, key) in enumerate(clases):
        # Calcula la posición Y del botón actual
        # Cada botón está separado 10 píxeles verticalmente
        # Editable: SÍ - Cambiar 10 para más/menos espacio entre botones
        y = start_y_clases + i * (btn_h + 10)
        # Añade el botón a la lista con su clave (key) y objeto Boton
        botones_clases.append((key, Boton(ANCHO // 2 - btn_w // 2, y, btn_w, btn_h, label)))
    
    # Botón "Volver atrás" en la pantalla de selección de clase
    # Posición: debajo del último botón de clase
    # Editable: SÍ - Puedes cambiar posición y tamaño
    boton_volver_seleccion = Boton(ANCHO // 2 - btn_w // 2, start_y_clases + len(clases) * (btn_h + 10) + 20, btn_w, btn_h, "Volver atrás")

    # ==================== ESTADÍSTICAS BASE POR CLASE ====================
    # Qué: Diccionario que almacena (Agilidad, Vitalidad) para cada clase
    # Para qué: Asignar estadísticas automáticamente cuando se crea un personaje
    # Cómo: Cada clase tiene dos números: agilidad y vitalidad
    # Editable: SÍ - Puedes cambiar los números para balancear clases
    # Nota: Las claves DEBEN coincidir con los 'key' en la lista 'clases'
    stats = {
        'clerigo': (5, 12),      # Clérigo: Agilidad=5, Vitalidad=12 (tanque de apoyo)
        'guerrero': (7, 15),     # Guerrero: Agilidad=7, Vitalidad=15 (tanque ofensivo)
        'mago': (8, 8),          # Mago: Agilidad=8, Vitalidad=8 (equilibrado)
        'luchador': (10, 10),    # Luchador: Agilidad=10, Vitalidad=10 (ágil)
    }

    # ==================== ESTADO Y VARIABLES ====================
    # Qué: Variables que rastrean el estado actual del programa
    # Para qué: Saber en qué pantalla está el usuario y qué personajes ha creado
    # Editable: NO - Se actualizan automáticamente en el bucle principal
    
    ejecutando = True                    # Flag: True si el juego debe continuar
    en_menu_principal = True             # Flag: True si estamos en menú principal
    en_pantalla_a = False                # Flag: True si estamos en pantalla de crear personaje
    en_seleccion_clase = False           # Flag: True si estamos seleccionando clase
    en_pantalla_clase = False            # Flag: True si estamos en pantalla de confirmación de clase
    personajes_creados = []              # Lista: almacena todos los personajes creados (máx 6)
    MAX_PERSONAJES = 6                   # Constante: máximo número de personajes permitidos
    clase_seleccionada = None            # Variable: la clase que el usuario está seleccionando
    
    # ==================== BOTONES PARA PANTALLA DE CLASE ====================
    # Qué: Botones específicos para la pantalla de confirmación de clase
    # Para qué: Permitir confirmar o volver en la pantalla de clase seleccionada
    # Editable: SÍ - Puedes cambiar posiciones, tamaños, textos
    boton_confirmar_clase = Boton(ANCHO // 2 - 150, ALTO - 120, 300, 60, "Confirmar")
    boton_volver_pantalla_clase = Boton(ANCHO // 2 - 150, ALTO - 50, 300, 60, "Volver atrás")

    # ==================== BUCLE PRINCIPAL ====================
    # Qué: Loop infinito que mantiene el juego corriendo
    # Para qué: Procesar eventos, actualizar lógica y dibujar cada fotograma
    # Cómo: Se ejecuta ~60 veces por segundo (según FPS)
    # Editable: NO - La estructura es correcta y necesaria
    while ejecutando:
        # Controla la velocidad del juego (60 FPS)
        # Qué: Limita la velocidad del bucle a FPS fotogramas por segundo
        # Para qué: Que el juego corra a velocidad constante
        # Editable: NO - Necesario para control de velocidad
        reloj.tick(FPS)
        
        # Obtiene la posición actual del ratón en la pantalla
        # Qué: Tupla (x, y) con la posición del ratón
        # Para qué: Detectar hover y clicks en botones/cuadrados
        # Editable: NO - Se obtiene automáticamente
        pos_mouse = pygame.mouse.get_pos()

        # ==================== MANEJO DE EVENTOS ====================
        # Qué: Procesa todos los eventos que ha generado el usuario (clicks, cerrar ventana, etc.)
        # Para qué: Responder a las acciones del usuario
        # Editable: SÍ - Puedes añadir más eventos (teclado, tecla ESC, etc.)
        for evento in pygame.event.get():
            # Detecta si el usuario cerró la ventana
            # Qué: pygame.QUIT se dispara cuando el usuario hace click en X
            # Para qué: Detener el juego correctamente
            # Editable: NO - Es estándar en pygame
            if evento.type == pygame.QUIT:
                ejecutando = False

            # Detecta clicks del ratón (botón izquierdo)
            # Qué: pygame.MOUSEBUTTONDOWN se dispara cuando el usuario hace click
            # evento.button == 1: verifica que sea el botón izquierdo (1=izq, 2=centro, 3=derecha)
            # Para qué: Saber cuándo el usuario hace click
            # Editable: SÍ - Puedes cambiar a evento.button == 3 para botón derecho, etc.
            if evento.type == pygame.MOUSEBUTTONDOWN and evento.button == 1:
                # --- MENÚ PRINCIPAL: Eventos de los botones principales ---
                # Qué: Procesa clicks en el menú principal
                # Para qué: Permitir al usuario ir a crear personaje o salir
                # Editable: SÍ - Puedes añadir más botones o cambiar la lógica
                if en_menu_principal:
                    # Click en botón "Personaje"
                    if boton_personaje.fue_clickeado(pos_mouse):
                        # Verifica que haya espacio para más personajes
                        if len(personajes_creados) < MAX_PERSONAJES:
                            # Cambia a pantalla A (crear personaje)
                            en_menu_principal = False
                            en_pantalla_a = True
                    # Click en botón "Salir del juego"
                    if boton_salir.fue_clickeado(pos_mouse):
                        # Cierra el juego
                        ejecutando = False

                # --- PANTALLA A: Eventos de los cuadrados y botones ---
                # Qué: Procesa clicks en los 6 cuadrados o en "Volver atrás"
                # Para qué: Permitir al usuario seleccionar un slot para crear personaje
                # Editable: SÍ - Puedes cambiar la lógica (ej. hacer que click en ocupado abre edición)
                elif en_pantalla_a:
                    # Loop sobre cada cuadrado para detectar clicks
                    for i, cuadro in enumerate(cuadros_rects):
                        # Verifica que sea un cuadrado vacío (índice i sin personaje)
                        if i >= len(personajes_creados) and cuadro.collidepoint(pos_mouse):
                            # Solo permite crear si hay espacio (debería ser redundante con el anterior check)
                            if len(personajes_creados) < MAX_PERSONAJES:
                                # Va a seleccionar clase
                                en_pantalla_a = False
                                en_seleccion_clase = True
                            break

                    # Click en botón "Volver atrás"
                    if boton_volver_pantalla_a.fue_clickeado(pos_mouse):
                        # Regresa al menú principal sin crear nada
                        en_pantalla_a = False
                        en_menu_principal = True

                # --- SELECCIÓN DE CLASE: Eventos de los botones de clase ---
                # Qué: Procesa clicks en los botones de las 4 clases disponibles
                # Para qué: Permitir al usuario seleccionar qué clase quiere
                # Editable: SÍ - Puedes añadir más clases o cambiar el orden
                elif en_seleccion_clase:
                    # Loop sobre cada botón de clase
                    for key, btn in botones_clases:
                        # Detecta click en un botón de clase
                        if btn.fue_clickeado(pos_mouse):
                            # Guarda la clase seleccionada (ej. "Clerigo")
                            clase_seleccionada = key.capitalize()
                            # Va a la pantalla de confirmación
                            en_seleccion_clase = False
                            en_pantalla_clase = True
                            break

                    # Click en botón "Volver atrás" regresa a pantalla A
                    if boton_volver_seleccion.fue_clickeado(pos_mouse):
                        en_seleccion_clase = False
                        en_pantalla_a = True

                # --- PANTALLA DE CLASE: Eventos de los botones de confirmación ---
                # Qué: Procesa clicks en "Confirmar" o "Volver atrás"
                # Para qué: Crear el personaje o volver a seleccionar clase
                # Editable: SÍ - Puedes añadir más opciones antes de confirmar
                elif en_pantalla_clase:
                    # Click en botón "Confirmar"
                    if boton_confirmar_clase.fue_clickeado(pos_mouse):
                        # Verifica que haya clase seleccionada y espacio disponible
                        if clase_seleccionada and len(personajes_creados) < MAX_PERSONAJES:
                            # Obtiene las estadísticas de la clase
                            agi, vit = stats[clase_seleccionada.lower()]
                            # Obtiene la imagen de la clase (o None si no existe)
                            imagen = imagenes_clases.get(clase_seleccionada.lower())
                            # Crea un nuevo personaje con imagen
                            nuevo = Personaje(clase_seleccionada, agi, vit, imagen)
                            # Lo añade a la lista
                            personajes_creados.append(nuevo)
                        # Regresa a pantalla A (6 cuadrados)
                        en_pantalla_clase = False
                        en_pantalla_a = True
                        clase_seleccionada = None

                    # Click en botón "Volver atrás"
                    if boton_volver_pantalla_clase.fue_clickeado(pos_mouse):
                        # Regresa a selección de clase sin crear
                        en_pantalla_clase = False
                        en_seleccion_clase = True
                        clase_seleccionada = None

        # ==================== ACTUALIZACIÓN DE BOTONES (HOVER) ====================
        # Qué: Actualiza el color de los botones según si el ratón está sobre ellos
        # Para qué: Dar feedback visual al usuario (sabe qué botones puede presionar)
        # Editable: NO - La lógica es correcta y necesaria
        if en_menu_principal:
            boton_personaje.actualizar(pos_mouse)
            boton_salir.actualizar(pos_mouse)
        elif en_pantalla_a:
            boton_volver_pantalla_a.actualizar(pos_mouse)
            # Detecta hover en cuadros vacíos
            cuadro_hover_index = -1
            for i, cuadro in enumerate(cuadros_rects):
                # Solo cambia color si el cuadro está vacío
                if i >= len(personajes_creados) and cuadro.collidepoint(pos_mouse):
                    cuadro_hover_index = i
                    break
        elif en_seleccion_clase:
            for _, btn in botones_clases:
                btn.actualizar(pos_mouse)
            boton_volver_seleccion.actualizar(pos_mouse)
        elif en_pantalla_clase:
            boton_confirmar_clase.actualizar(pos_mouse)
            boton_volver_pantalla_clase.actualizar(pos_mouse)

        # ==================== DIBUJO DE PANTALLA ====================
        # Qué: Renderiza todos los elementos gráficos en la pantalla
        # Para qué: Mostrar la interfaz del juego al usuario
        # Cómo: Primero limpia la pantalla, luego dibuja elementos según la pantalla actual
        # Editable: SÍ - Puedes cambiar colores, posiciones, textos, etc.
        
        # Limpia la pantalla con color negro
        # Qué: Llena toda la pantalla con color NEGRO
        # Para qué: Borrar el fotograma anterior (necesario antes de dibujar uno nuevo)
        # Editable: SÍ - Puedes cambiar a otro color (ej. GRIS)
        pantalla.fill(NEGRO)

        # --- MENÚ PRINCIPAL ---
        # Qué: Dibuja la pantalla del menú principal
        # Para qué: Mostrar al usuario la pantalla inicial con sus botones
        # Editable: SÍ - Puedes cambiar textos, posiciones, colores
        if en_menu_principal:
            # Renderiza el título "Menú Principal"
            titulo = fuente_titulo.render("Menú Principal", True, BLANCO)
            # Calcula la posición para centrar el título horizontalmente, 100 píxeles desde arriba
            # Editable: SÍ - Cambiar 100 para mover título arriba/abajo
            titulo_rect = titulo.get_rect(center=(ANCHO // 2, 100))
            # Dibuja el título en la pantalla
            pantalla.blit(titulo, titulo_rect)

            # Dibuja los dos botones del menú principal
            boton_personaje.dibujar(pantalla)
            boton_salir.dibujar(pantalla)

        # --- PANTALLA A (6 Cuadrados con imágenes o +) ---
        # Qué: Dibuja la pantalla para crear personaje (6 cuadrados)
        # Para qué: Mostrar al usuario los slots disponibles y cuáles están ocupados
        # Editable: SÍ - Puedes cambiar textos, colores, posiciones
        elif en_pantalla_a:
            # Renderiza el título "Crear Personaje"
            titulo = fuente_titulo.render("Crear personaje", True, BLANCO)
            # Posición: centrado horizontalmente, 80 píxeles desde arriba
            # Editable: SÍ - Cambiar 80 para mover título
            titulo_rect = titulo.get_rect(center=(ANCHO // 2, 80))
            pantalla.blit(titulo, titulo_rect)

            # Loop que dibuja cada uno de los 6 cuadrados
            # Qué: Itera sobre cada cuadrado y lo dibuja con su estado (ocupado o vacío)
            # Para qué: Mostrar visualmente los 6 slots al usuario
            # Editable: SÍ - Puedes cambiar colores, tamaño, estilo
            for i, cuadro in enumerate(cuadros_rects):
                if i < len(personajes_creados):
                    # CUADRO OCUPADO: dibuja el personaje que está en este slot
                    # Qué: Si el índice i tiene un personaje, lo dibuja
                    # Para qué: Mostrar que este slot está ocupado
                    color_cuadro = AZUL
                    # Dibuja rectángulo relleno
                    pygame.draw.rect(pantalla, color_cuadro, cuadro)
                    # Dibuja borde blanco alrededor
                    # Editable: SÍ - Cambiar 3 para borde más grueso/fino
                    pygame.draw.rect(pantalla, BLANCO, cuadro, 3)
                    
                    # Intenta mostrar la imagen del personaje
                    # Qué: Si existe imagen, la dibuja; si no, muestra la inicial
                    # Para qué: Mostrar representación visual del personaje
                    # Editable: SÍ - Puedes cambiar la lógica para mostrar siempre la inicial
                    if personajes_creados[i].imagen:
                        # Dibuja la imagen escalada en el cuadro
                        pantalla.blit(personajes_creados[i].imagen, cuadro.topleft)
                    else:
                        # Dibuja la inicial de la clase (C para Clérigo, G para Guerrero, etc.)
                        inicial = fuente_boton.render(personajes_creados[i].clase[0], True, BLANCO)
                        pantalla.blit(inicial, inicial.get_rect(center=cuadro.center))
                else:
                    # CUADRO VACÍO: dibuja imagen o "+" en el slot vacío
                    # Qué: Si el índice i NO tiene personaje, dibuja imagen o "+"
                    # Para qué: Mostrar que este slot está disponible y es clickeable
                    # Editable: SÍ - Puedes cambiar colores, símbolo (usar imagen, icono, etc.)
                    # Cambia color si el ratón está sobre el cuadro (hover)
                    color_cuadro = AZUL_CLARO if cuadro_hover_index == i else AZUL
                    # Dibuja rectángulo relleno
                    pygame.draw.rect(pantalla, color_cuadro, cuadro)
                    # Dibuja borde blanco
                    pygame.draw.rect(pantalla, BLANCO, cuadro, 3)
                    
                    # Dibuja imagen si existe, si no dibuja "+"
                    # Qué: Muestra imagen_vacio si existe, sino símbolo "+"
                    # Para qué: Estética (imagen mejor que símbolo)
                    # Editable: SÍ - Cambiar lógica para siempre mostrar "+"
                    if imagen_vacio:
                        pantalla.blit(imagen_vacio, cuadro.topleft)
                    else:
                        # Dibuja símbolo "+"
                        # Editable: SÍ - Puedes cambiar "+" a otro símbolo (ej. "?", "NEW", etc.)
                        plus = fuente_titulo.render("+", True, BLANCO)
                        pantalla.blit(plus, plus.get_rect(center=cuadro.center))

            # Dibuja el botón "Volver atrás"
            boton_volver_pantalla_a.dibujar(pantalla)

            # Dibuja contador de personajes en la esquina superior izquierda
            # Qué: Muestra "Personajes: X/6" para que el usuario vea cuántos tiene
            # Para qué: Información visual clara del progreso
            # Editable: SÍ - Puedes cambiar texto, posición (20, 20), color
            contador = fuente_texto.render(f"Personajes: {len(personajes_creados)}/{MAX_PERSONAJES}", True, BLANCO)
            pantalla.blit(contador, (20, 20))

        # --- SELECCIÓN DE CLASE ---
        # Qué: Dibuja la pantalla donde el usuario elige qué clase quiere
        # Para qué: Mostrar al usuario las 4 opciones de clase
        # Editable: SÍ - Puedes cambiar título, colores, orden de clases
        elif en_seleccion_clase:
            # Renderiza título "Escoger clase"
            # Editable: SÍ - Puedes cambiar el texto
            titulo = fuente_titulo.render("Escoger clase", True, BLANCO)
            # Posición: centrado, 80 píxeles desde arriba
            # Editable: SÍ - Cambiar 80 para mover título
            titulo_rect = titulo.get_rect(center=(ANCHO // 2, 80))
            pantalla.blit(titulo, titulo_rect)

            # Dibuja los 4 botones de clase
            for _, btn in botones_clases:
                btn.dibujar(pantalla)

            # Dibuja el botón "Volver atrás"
            boton_volver_seleccion.dibujar(pantalla)

        # --- PANTALLA DE CLASE (Confirmación) ---
        # Qué: Dibuja la pantalla donde aparece el nombre de la clase seleccionada en grande
        # Para qué: Confirmar la selección del usuario y mostrar stats
        # Editable: SÍ - Puedes cambiar textos, posiciones, agregar más información
        elif en_pantalla_clase and clase_seleccionada:
            # Renderiza el nombre de la clase seleccionada como título principal
            # Qué: Muestra el nombre de la clase (Clérigo, Guerrero, Mago, Luchador)
            # Para qué: Que el usuario confirme visualmente qué clase eligió
            # Editable: SÍ - Puedes cambiar tamaño de fuente, color, posición
            titulo = fuente_titulo.render(clase_seleccionada, True, BLANCO)
            # Posición: centrado horizontalmente, 150 píxeles desde arriba
            # Editable: SÍ - Cambiar 150 para mover título
            titulo_rect = titulo.get_rect(center=(ANCHO // 2, 150))
            pantalla.blit(titulo, titulo_rect)

            # Obtiene las estadísticas de la clase seleccionada
            agi, vit = stats[clase_seleccionada.lower()]
            # Renderiza texto con estadísticas
            # Qué: Muestra "Agilidad: X   Vitalidad: Y" para que vea qué stats tiene
            # Para qué: Información para tomar la decisión de confirmar
            # Editable: SÍ - Puedes cambiar formato, agregar más stats, cambiar posición (280)
            stats_text = fuente_texto.render(f"Agilidad: {agi}   Vitalidad: {vit}", True, BLANCO)
            pantalla.blit(stats_text, stats_text.get_rect(center=(ANCHO // 2, 280)))

            # Dibuja los botones "Confirmar" y "Volver atrás"
            boton_confirmar_clase.dibujar(pantalla)
            boton_volver_pantalla_clase.dibujar(pantalla)

        # ==================== ACTUALIZACIÓN DE PANTALLA ====================
        # Qué: Actualiza la pantalla con todo lo que se acaba de dibujar
        # Para qué: Mostrar los cambios al usuario (sin esto no vería nada nuevo)
        # Editable: NO - Es obligatorio y debe estar al final de cada iteración
        pygame.display.flip()

    # ==================== CIERRE DE PYGAME ====================
    # Qué: Cierra pygame de manera ordenada
    # Para qué: Liberar recursos y salir del programa correctamente
    # Editable: NO - Debe estar al final para cerrar todo correctamente
    pygame.quit()
    sys.exit()


# ==================== PUNTO DE ENTRADA DEL PROGRAMA ====================
# Qué: Verifica que el archivo se ejecuta directamente (no se importa como módulo)
# Para qué: Permitir que el script se ejecute como programa principal
# Editable: NO - Esta es la forma estándar en Python
if __name__ == "__main__":
    main()
```
